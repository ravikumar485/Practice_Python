```python
#variable length arguments
def f(*args):
    print(type(args)) #tupple
    sum=0
    for i in args:
        sum=sum+i
    return sum

print(f(10,20,30,40,50))
```

```python
def f(*n):
    sum=0
    for i in n:
        sum=sum+i

print(f(1,2,3,4,5))
```

```python
def f(n,*r):
    print(n, r)

f(1,2,3,4,5)
```

```python
def f(*r,n):
    print(n, r)

f(1,2,3,4,5) #f() missing 1 required keyword-only argument: 'n'
```

```python
def f(*r,n):
    print(n, r)

f(2,3,4,5,n=10) 
```
